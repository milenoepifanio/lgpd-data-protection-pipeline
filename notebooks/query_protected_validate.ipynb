```python
from pathlib import Path
import sys

project_root = Path.cwd().resolve()
while project_root != project_root.parent and not (project_root / "utils").is_dir():
    project_root = project_root.parent
sys.path.insert(0, str(project_root))

from utils.config import PROTECTED_CUSTOMERS_PATH
from utils.dataset import load_dataset
```

```python
def main() -> None:
    dataframe = load_dataset(PROTECTED_CUSTOMERS_PATH)

    print(f"Arquivo: {PROTECTED_CUSTOMERS_PATH}")
    print(f"Registros: {len(dataframe):,}")
    print(f"Colunas: {len(dataframe.columns)}")
    print()
    print("Nomes das colunas:")
    print(", ".join(dataframe.columns))
    print()
    print("Primeiras 5 linhas:")
    print(dataframe.head(5).to_string(index=False))
```

```python
if __name__ == "__main__":
    main()
```
