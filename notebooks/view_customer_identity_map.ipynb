# Consulta do Registro Restrito de Clientes

Este notebook consulta exclusivamente o arquivo `data/restricted/customer_identity_map.parquet`.

O arquivo contém dados pessoais e sensíveis. Execute este notebook somente em ambiente autorizado e não compartilhe suas saídas.

```python
from pathlib import Path

import pandas as pd

project_root = Path.cwd().resolve()
while project_root != project_root.parent and not (project_root / "data").is_dir():
    project_root = project_root.parent

restricted_path = (
    project_root
    / "data"
    / "restricted"
    / "customer_identity_map.parquet"
)

print(f"Arquivo: {restricted_path}")
assert restricted_path.exists(), f"Arquivo não encontrado: {restricted_path}"
```

```python
customer_identity_map = pd.read_parquet(restricted_path)

print(f"Registros: {len(customer_identity_map):,}")
print(f"Colunas: {len(customer_identity_map.columns)}")
print("\nNomes das colunas:")
print(list(customer_identity_map.columns))

print("\nTipos das colunas:")
display(customer_identity_map.dtypes.rename("tipo").to_frame())
```

```python
pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)
pd.set_option("display.max_colwidth", 80)

display(customer_identity_map.head(10))
```
